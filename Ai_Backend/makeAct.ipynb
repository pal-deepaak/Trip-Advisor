```python
def rebuild_activity(row):
    name = str(row["destination"]).lower()

    categories = {
        "trekking": ["trek", "hill", "valley", "peak", "trail"],
        "beach": ["beach"],
        "religious": ["temple", "mandir", "dham", "gurudwara", "mosque"],
        "photography": ["park", "sanctuary", "wildlife", "lake"],
        "adventure": ["rafting", "adventure", "camp", "ski"]
    }

    for category, keywords in categories.items():
        if any(word in name for word in keywords):
            return category

    return "general"
```
