```python
def calculate_score(row, user):

    score = 0

    if row["type"] == user["type"]:
        score += 3

    if row["budget_category"] == user["budget"]:
        score += 2

    if row["best_season"] == user["season"]:
        score += 2

    if row["region"] == user["region"]:
        score += 2

    if user["activities"] in row["activities"]:
        score += 3
        
    score += row["rating"] * 0.2
    score += row["popularity_score"] * 0.02

    return score
```

```python
user_input = {
    "type": "adventure",
    "budget": "high",
    "season": "summer",
    "activities": "trekking",
    "region": "North"
}
```

```python
df['score'] = df.apply(lambda row : calculate_score(row, user_input), axis=1)
```

```python
extra = df[["state","region"]].drop_duplicates()
extra.rename(columns={"state" : "destination"},inplace=True)
df = pd.concat([df,extra],ignore_index=True)
df
```

```python
def recommend_cities(data, user):

    # filter by region
    filtered = data[data["region"] == user["region"]]

    # filter by type
    filtered = filtered[filtered["type"] == user["type"]]

    # अगर empty हो गया
    if len(filtered) == 0:
        filtered = data[data["region"] == user["region"]]

    # top states निकालो
    top_states = (
        filtered.groupby("state")
        .size()
        .sort_values(ascending=False)
        .head(3)
        .index
        .tolist()
    )

    return top_states
```

```python
selected_state = "himachal Pradesh"
```

```python
def recommended_spots(df,user,selected_state):

    filtered = df[df['state'] == selected_state]

    filtered['score'] = filtered.apply(
        lambda row : calculate_score(row,user), axis=1
    )

    top_spots = filtered.sort_values(by="score",ascending=False).head()

    return top_spots[["destination", "score"]]
```

```python


cities = recommend_cities(df,user_input)
print("Suggested Cities : ",cities)

selected  = cities[0]

spots = recommended_spots(df,user_input,selected)
print(spots)
```
