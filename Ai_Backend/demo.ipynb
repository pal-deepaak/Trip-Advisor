```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('places.csv',encoding='latin1')
data.head()
```

```python
data['state'].unique()
```

```python
data['state'].unique()
```

```python

data['state'] = data['state'].str.lower().str.strip()
data['state'].unique().shape
```

```python
data['state'].duplicated().sum()
```

```python
state_mapping = {
    "maharastra": "maharashtra",
    "maharahtra": "maharashtra",
    "gujrat": "gujarat",
    "karanataka": "karnataka",
    "kerala ": "kerala"
}

data['state'] = data['state'].replace(state_mapping)
data['state'].unique()
```

```python
data.rename(columns={'popular_destination' : 'destination'},inplace=True)
```

```python
def get_regions(state):
    state = str(state).lower().strip()
    north = ["himachal pradesh", "uttarakhand", "punjab", "jammu and kashmir", "ladakh", "haryana", "delhi"]
    south = ["kerala", "tamil nadu", "karnataka", "andhra pradesh", "telangana"]
    east = ["west bengal", "odisha", "bihar", "jharkhand"]
    west = ["rajasthan", "gujarat", "goa", "maharashtra"]
    northeast = ["assam", "meghalaya", "sikkim", "nagaland", "tripura", "arunachal pradesh"]

    if state in north: 
        return "North"
    elif state in south:
        return "South"
    elif state in east:
        return "East"
    elif state in west:
        return "West"
    elif state in northeast:
        return "North-East"
    else:
        return "Unknown"


data['region'] = data['state'].apply(get_regions)
   
```

```python
data['region'].unique()
```

```python
data['interest'] = data['interest'].str.replace("\t","")
data['interest'] = data['interest'].str.strip().str.lower()
```

```python
data['interest_list'] = data['interest'].str.split(',')
```

```python
data.head()
```

```python
def assign_type(interest_list):
    for interest in interest_list:
        interest = interest.strip()

        if "beach" in interest:
            return "beach"
        elif "adventure" in interest:
            return "adventure"
        elif "wildlife" in interest or "natural" in interest:
            return "wildlife"
        elif "heritage" in interest or "cultural" in interest:
            return "heritage"
        elif "religious" in interest or "spirtual" in interest:
            return "religious"
        
    return "city"
```

```python
data['type'] = data['interest_list'].apply(assign_type)
```

```python
data[['interest','interest_list','type','destination']].head(10)
```

```python
def assign_activities(type):

    if type == "mountain":
        return "trekking,snow,adventure,viewpoints"

    elif type == "beach":
        return "beach,water sports,nightlife,relaxation"

    elif type == "heritage":
        return "culture,forts,shopping,history"

    elif type == "religious":
        return "temple,spiritual,peace"

    elif type == "wildlife":
        return "safari,photography,nature"

    elif type == "adventure":
        return "trekking,rafting,paragliding"

    else:
        return "exploration,local experience"
```

```python
data["activities"] = data["type"].apply(assign_activities)
```

```python
data.head()
```

```python
def assign_budget(price):

    if price == 0:
        return "low"

    elif price < 50:
        return "low"

    elif price < 200:
        return "medium"

    elif price < 500:
        return "high"

    else:
        return "luxury"
```

```python
data["budget_category"] = data["price_fare"].apply(assign_budget)
```

```python
data['budget_category'].unique()
```

```python
data['type'].unique()
```

```python
def assign_season(type):

    if type == "mountain" or type == "adventure":
        return "summer"
    
    elif type == "beach" or type == "desert" or type == "wildlife":
        return "winter"
    
    elif type == "heritage":
        return "winter"   # sightseeing comfortable

    elif type == "religious":
        return "all" 
    
    else:
        return "All"
```

```python
data['best_season'] = data['type'].apply(assign_season)
```

```python
data['rating'] = data['google_rating']
```

```python
data['popularity_score'] = np.random.randint(50,100, size = len(data))
```

```python
def generate_description(row):
    return f"{row['destination']} is a {row['type']} destination in {row['state']} known for {row['activities']}."

```

```python
def get_user_type(type):

    if type == "adventure":
        return "adventure lover"
    
    elif type == "wildlife":
        return "history and cultural lover"
    
    elif type == "heritage":
        return "history and culture lovers"

    elif type == "religious":
        return "spiritual seekers"

    elif type == "city":
        return "explorers and tourists"

    else:
        return "travelers"
```

```python
def generate_desc(row):

    user_type = get_user_type(row['type'])

    return f"{row['destination']} is a {row['type']} destination in {row['state']} known for {row['activities']}. It is ideal for {user_type}"
```

```python
data['description'] = data.apply(generate_desc,axis=1)
```

```python
data = data.drop(columns=["interest_list","google_rating","price_fare","city"])
```
