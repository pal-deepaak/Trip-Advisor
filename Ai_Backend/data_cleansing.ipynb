```python
import pandas as pd
data = pd.read_csv("proj_data.csv")
```

```python
for col in ["type","budget_category","best_season","region","activities"]:
    data[col] = data[col].astype(str).str.lower()
data['region'].value_counts()
```

```python
df = data[data['region'] != "unknown"]
df['region'].value_counts()
```

```python
df.shape
```

```python
df = df.dropna(subset=["city", "type", "activities"])
df.shape
```

```python
df["activities"] = df["activities"].apply(
    lambda x: " ".join([i.strip() for i in str(x).split(",")])
)
```

```python
df["activities"] = df["activities"].str.replace(r"\s+", " ", regex=True)
```

```python
df = df[df['popularity_score'] > 50]
df = df[(df["rating"] > 3.5) & (df["rating"] < 5.0)]
df.shape
```

```python
invalid_cities = [
    "sheffield",
    "pudupakkam",
    "eguvamamidi",
    "khinnanavli frh"
]

df = df[~df["city"].str.lower().isin(invalid_cities)]
df.shape
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df.to_csv("shi_data.csv",index=False)
```

```python
def rebuild_activity(row):
    name = str(row["destination"]).lower()
    interest = str(row["interest"]).lower()
    type_ = str(row["type"]).lower()

    text = name + " " + interest + " " + type_

    # 🔥 trekking
    if any(word in text for word in ["trek", "trail", "peak", "mountain"]):
        return "trekking"

    # 🔥 beach
    elif any(word in text for word in ["beach", "sea", "coast"]):
        return "beach"

    # 🔥 religious
    elif any(word in text for word in ["temple", "mandir", "dham", "gurudwara", "mosque", "spiritual"]):
        return "religious"

    # 🔥 wildlife / nature
    elif any(word in text for word in ["wildlife", "sanctuary", "forest", "nature", "park"]):
        return "wildlife"

    # 🔥 photography
    elif any(word in text for word in ["valley", "lake", "waterfall", "falls"]):
        return "photography"

    # 🔥 adventure
    elif any(word in text for word in ["adventure", "rafting", "paragliding", "camp", "ski"]):
        return "adventure"

    # 🔥 heritage
    elif any(word in text for word in ["fort", "museum", "heritage", "palace", "history"]):
        return "heritage"

    # 🔥 shopping (NEW)
    elif any(word in text for word in ["market", "shopping", "bazaar"]):
        return "shopping"

    # 🔥 city / exploration (NEW)
    elif any(word in text for word in ["city", "exploration", "sightseeing"]):
        return "exploration"

    else:
        return "general"
```

```python
df["activities"] = df.apply(rebuild_activity, axis=1)
```
